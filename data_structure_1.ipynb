Question:

Given an array of integers, find the largest element in the array.

Example:

Input:  [12, 5, 8, 21, 3, 17]

Output: 21

```python
def find_largest(arr):
    if not arr:
        return None
    largest_no = arr[0]
    for num in arr:
        if num > largest_no:
            largest_no = num
    return largest_no

arr=find_largest([12, 5, 8, 21, 3, 17])
print(arr)
```

Question: Find the second largest element

Given:

arr = [10, 5, 8, 20, 15, 20]


Expected output:

```python
def second_largest_no(arr):
    if len(arr) < 2 or not arr:
        return None 

    lagest_no=[0]
    second_largest_no=[0]
    for num in arr:
        if num > lagest_no[0]:
            lagest_no[0] = num
    for num in arr:
        if num > second_largest_no[0] and num < lagest_no[0]:
            second_largest_no[0] = num
    return second_largest_no[0]


arr=second_largest_no([10, 5, 8, 20, 15, 20])
print(arr)
```
