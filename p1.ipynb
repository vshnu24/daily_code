***amicable numbers***

```python
n1=int(input())
n2=int(input())
ami1=0
ami2=0

for i in range (1,n1):
   if n1 % i ==0:
      ami1+=i
for i in range (1,n2):
   if n2 % i ==0:
      ami2+=i
if ami1==n2 and ami2==n1:
   print(f"{n1} and {n2} are amicable numbers")
else:
   print(f"{n1} and {n2} are not amicable numbers")
```

***strong number***

```python
n=int(input())
temp=n
total=0
while n!=0:
    rem=n%10
    fact=1
    for i in range(1,rem+1):
        fact=fact*i
    total+=fact
if temp==total:
    print("strong number")
else:
    print("not strong")

```
