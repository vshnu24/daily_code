***leap year***

```python
year=int(input("Enter a year: "))
if year%4==0 and year%100!=0 or year%400==0:
    print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
```

```python
u=int(input())
cnt=0
for i in range (1,u+1):
    if u %i ==0:
        cnt+=1
if cnt == 2:
    print(f"{u} is a prime number")
else:
    print(f"{u} is not a prime number")
```

***gcd***

```python
u1=int(input())
u2=int(input())
gcd=1
for i in range (1,u1+1):
    if u1 % i ==0 and u2 % i ==0:
        gcd=i;
print(f"GCD of {u1} and {u2} is {gcd}")
```

***amicable numbers***

```python
n1=int(input())
n2=int(input())
ami1=0
ami2=0

for i in range (1,n1):
   if n1 % i ==0:
      ami1+=i
for i in range (1,n2):
   if n2 % i ==0:
      ami2+=i
if ami1==n2 and ami2==n1:
   print(f"{n1} and {n2} are amicable numbers")
else:
   print(f"{n1} and {n2} are not amicable numbers")
```
